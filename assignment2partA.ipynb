```python
import pandas as pd
import numpy as np


a=pd.read_csv('age_gender.csv')
print(a.head())

b=pd.read_excel('Cola.xlsx')
print(b.head())
```

```python
print("types:",type(a),type(b))
print("csv Columns:",a.columns)
print("Excel Columns:",b.columns)
print("csv index:",a.index.tolist())
print("Excel index:",b.index.tolist())
```

```python
print("first three rows of csv file:\n",a.head(3))
print("\n")
print("first three rows of excel file:\n",b.head(3))
print("\n")

print("last three rows of csv file:\n",a.tail(3))
print("\n")
print("last three rows of excel file:\n",b.tail(3))
print("\n")

print("csv file info:\n")
a.info()
print("\n")
print("excel file info:\n")
b.info()
print("\n")

print("csv file description:\n",a.describe())
print("excel file description:\n",b.describe())

print("csv file shape:",a.shape)
print("excel file shape:",b.shape)
```

```python
print("loc example csv file:\n",a.loc[0:2,['age','gender']])
print("loc example excel file:\n",b.loc[0:2,['Cola','Price']])
print("\n")
print("iloc example csv file:\n",a.iloc[0:2,[0,1]])
print("iloc example excel file:\n",b.iloc[0:2,[0,1]])
print("\n")
print("\n Boolean indexing example csv file:\n",a[a['age']>30])
print("\n Boolean indexing example excel file:\n",b[b['Price']>1.5])
```

```python
print("\n sum of age column in csv file:")
display(a.groupby('gender')['age'].sum())

print("\n mean of age column in csv file:")
display(a.groupby('gender')['age'].mean())

print("\n multiple aggregation functions by gender in csv file:")
display(a.groupby('gender')['age'].agg([np.sum, np.mean, np.std]))
```
